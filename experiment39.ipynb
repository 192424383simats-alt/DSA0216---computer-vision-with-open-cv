```
import cv2
import numpy as np
from IPython.display import HTML
from base64 import b64encode

# Load uploaded car cascade file
car_cascade = cv2.CascadeClassifier("haarcascade_car.xml")

# Check if cascade classifier is loaded successfully
if car_cascade.empty():
    raise IOError('Could not load car cascade file. Make sure "haarcascade_car.xml" is uploaded and accessible.')

cap = cv2.VideoCapture("/content/sample_640x360.mp4")

frames = []

while True:
    ret, frame = cap.read()
    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    cars = car_cascade.detectMultiScale(gray, 1.1, 2)

    for (x,y,w,h) in cars:
        cv2.rectangle(frame, (x,y), (x+w,y+h), (0,0,255), 2)

    frames.append(frame)

cap.release()

# Save output video
height, width, layers = frames[0].shape
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter("vehicle_output.mp4", fourcc, 20.0, (width, height))

for frame in frames:
    out.write(frame)

out.release()

# Display video
video = open("vehicle_output.mp4", "rb").read()
video_encoded = b64encode(video).decode()

HTML(f"""
<video width=600 controls>
  <source src="data:video/mp4;base64,{video_encoded}" type="video/mp4">
</video>
""")
```
