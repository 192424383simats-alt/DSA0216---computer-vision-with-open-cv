```
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image in grayscale
img = cv2.imread("/content/Picture3.jpg", 0)

# High-boost constant
A = 2

# High-boost mask (4-neighbour)
kernel = np.array([[0, -1, 0],
                   [-1, A+4, -1],
                   [0, -1, 0]])

# Apply filter
highboost = cv2.filter2D(img, -1, kernel)

# Display images
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(highboost, cmap='gray')
plt.title("High-Boost Image")
plt.axis('off')

plt.show()
```
