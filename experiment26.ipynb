```
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("/content/Picture1.jpg")

# Convert BGR to RGB (important for matplotlib)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Add watermark text
cv2.putText(img_rgb, "WATERMARK", (50, 50),
            cv2.FONT_HERSHEY_SIMPLEX,
            1, (255, 0, 0), 2)

# Display image
plt.imshow(img_rgb)
plt.axis('off')
plt.title("Watermarked Image")
plt.show()

```
