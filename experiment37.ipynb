```
import cv2
import numpy as np
from IPython.display import HTML
from base64 import b64encode

# Read video
cap = cv2.VideoCapture("/content/sample_640x360.mp4")

frames = []

while True:
    ret, frame = cap.read()
    if not ret:
        break
    frames.append(frame)

cap.release()

# Get video properties
height, width, layers = frames[0].shape
fourcc = cv2.VideoWriter_fourcc(*'mp4v')

# Create output video
out = cv2.VideoWriter("reversed_video.mp4", fourcc, 20.0, (width, height))

# Write frames in reverse
for frame in reversed(frames):
    out.write(frame)

out.release()

# Display in Colab
video = open("reversed_video.mp4", "rb").read()
video_encoded = b64encode(video).decode()

HTML(f"""
<video width=600 controls>
  <source src="data:video/mp4;base64,{video_encoded}" type="video/mp4">
</video>
""")
```
