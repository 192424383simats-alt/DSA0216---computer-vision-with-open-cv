```
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Open video
cap = cv2.VideoCapture("/content/sample_640x360.mp4")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    rows, cols = frame.shape[:2]

    # Original 4 points (example)
    pts1 = np.float32([[100,100], [400,100], [100,400], [400,400]])

    # Destination 4 points
    pts2 = np.float32([[0,0], [cols,0], [0,rows], [cols,rows]])

    # Perspective matrix
    M = cv2.getPerspectiveTransform(pts1, pts2)

    # Apply transformation
    result = cv2.warpPerspective(frame, M, (cols, rows))

    cv2_imshow(frame)
    cv2_imshow(result)
```
