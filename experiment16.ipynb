```
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("/content/Picture1.jpg") # Corrected filename

if img is None:
    print("Error: Could not read the image. Please check the file path.")
else:
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    edges = cv2.Canny(gray, 100, 200)

    cv2_imshow(edges) # Using cv2_imshow for Colab compatibility

    # cv2.waitKey(0) # Not needed with cv2_imshow
    # cv2.destroyAllWindows() # Not needed with cv2_imshow
```
