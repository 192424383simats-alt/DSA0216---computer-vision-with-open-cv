```
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image in grayscale
img = cv2.imread("/content/Picture5.jpg", 0)

# Create 5x5 kernel
kernel = np.ones((5,5), np.uint8)

# Apply erosion
eroded = cv2.erode(img, kernel, iterations=1)

# Display
plt.imshow(eroded, cmap='gray')
plt.title("Erosion")
plt.axis('off')
plt.show()
```
