```
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read image
img = cv2.imread("/content/Picture2.jpg")

rows, cols = img.shape[:2]

# Source points
pts1 = np.float32([[50,50], [300,50], [50,300], [300,300]])

# Destination points
pts2 = np.float32([[0,0], [cols,0], [0,rows], [cols,rows]])

# Compute Homography (DLT method internally)
H, status = cv2.findHomography(pts1, pts2)

# Apply transformation
result = cv2.warpPerspective(img, H, (cols, rows))

# Show images
cv2_imshow(img)
cv2_imshow(result)

# cv2.waitKey(0) # Not needed with cv2_imshow
# cv2.destroyAllWindows() # Not needed with cv2_imshow
```
