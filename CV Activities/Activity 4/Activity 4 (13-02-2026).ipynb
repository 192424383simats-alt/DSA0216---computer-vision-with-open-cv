```
import cv2
import numpy as np
import matplotlib.pyplot as plt


img = cv2.imread("/content/p1.jpeg", 0)

level0 = img
level1 = cv2.pyrDown(level0)
level2 = cv2.pyrDown(level1)
level3 = cv2.pyrDown(level2)
level4 = cv2.pyrDown(level3)

images = [level0, level1, level2, level3, level4]
zoom_labels = ["High", "Med-High", "Medium", "Low", "Very Low"]


means = []
variances = []

for image in images:
    flat = image.flatten()
    means.append(np.mean(flat))
    variances.append(np.var(flat))

for i in range(len(images)):
    print(f"{zoom_labels[i]} Resolution")
    print("Mean     :", means[i])
    print("Variance :", variances[i])
    print("----------------------------")


plt.figure()
plt.plot(zoom_labels, variances, marker='o')
plt.xlabel("Zoom Level")
plt.ylabel("Variance of Pixel Intensity")
plt.title("Zoom Level vs Pixel Variance")
plt.show()
```
