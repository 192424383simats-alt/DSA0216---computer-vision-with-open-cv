```
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((400, 400), dtype=np.uint8)

cv2.circle(img, (120, 200), 15, 255, -1)

cv2.circle(img, (300, 200), 60, 255, -1)

blur_small = cv2.GaussianBlur(img, (5, 5), sigmaX=1)
log_small = cv2.Laplacian(blur_small, cv2.CV_64F)

blur_large = cv2.GaussianBlur(img, (21, 21), sigmaX=6)
log_large = cv2.Laplacian(blur_large, cv2.CV_64F)


plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image (Two Blobs)")
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(log_small, cmap='gray')
plt.title("Small Sigma (Scale = 1)")
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(log_large, cmap='gray')
plt.title("Large Sigma (Scale = 3)")
plt.axis('off')

plt.tight_layout()
plt.show()
```
